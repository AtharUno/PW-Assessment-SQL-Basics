Q1. Create a New Database and  Table for Employees
Task: Create a new database named and Create a table named with the following
columns:
Column Name   Data Type         Constraint
employee_id   INT               PRIMARY KEY
first_name    VARCHAR (50)
last_name     VARCHAR (50)
department    VARCHAR (50)
salary        INT
hire_date     DATE


  -Ans:

CREATE DATABASE company_db;
USE company_db;
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    department VARCHAR(50),
    salary INT,
    hire_date DATE
);

Q3. Display All Employee Records Sorted by Salary (Lowest to Highest)
Hint: Use the clause on the salary column.

-Ans:
SELECT * FROM employees
ORDER BY salary;

Q4. Show Employees Sorted by Department (A–Z) and Salary (High → Low)

-Ans: SELECT * FROM employees
ORDER BY department ASC, salary DESC;

Q5. List All Employees in the IT Department, Ordered by Hire Date (Newest First)

-Ans: SELECT * FROM employees
WHERE department = 'IT'
ORDER BY hire_date DESC;

Q9. What is the Difference Between a Primary Key and a Foreign Key?

-Ans:
A Primary Key is a unique key for all records in a table.
A Foreign Key connects one table to another by referencing the primary key of another table.

Q10. What Are Constraints in SQL and Why Are They Used?

-Ans:
Constraints are a set of rules applied on data columns in a table. Constraints help to fix the type of data that can go into a table range.

Why they are used:
1. Constraints ensure that the data remains accurate, reliable, and consistent over time.
2. Constraint does not allow to enter any wrong or blank cells. It helps in getting the data as per the set norms.
